# 01. Exploratory Data Analysis (EDA) - Data Riil BPS
## Analisis Efektivitas dan Ketepatan Sasaran Bantuan Sosial

Notebook ini memuat tahapan **Data Understanding** sesuai metodologi CRISP-DM:
1. Memuat dataset 514 Kabupaten/Kota se-Indonesia hasil scraping BPS.
2. Menelaah struktur data, tipe variabel, dan kelengkapan informasi.
3. Visualisasi distribusi indikator kemiskinan ($P_0, P_1, P_2$), pengeluaran, IPM, dan fasilitas perumahan.
4. Analisis korelasi antar-indikator kemiskinan dan alokasi bantuan sosial.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Set plot styling
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

# 1. Memuat Data
df = pd.read_csv('../data/raw/real_bps_kabupaten_kota.csv')
print(f'Dimensi Dataset: {df.shape[0]} Baris, {df.shape[1]} Kolom')
df.head()
```

```python
# 2. Ringkasan Statistik Deskriptif
print('=== RINGKASAN STATISTIK INDIKATOR UTAMA BPS ===')
cols_key = ['persentase_penduduk_miskin', 'garis_kemiskinan', 'indeks_kedalaman_p1', 
            'pengeluaran_per_kapita', 'indeks_pembangunan_manusia_ipm', 
            'persentase_sanitasi_layak', 'persentase_rumah_layak_huni', 'rasio_cakupan_bansos']
df[cols_key].describe().T
```

```python
# 3. Visualisasi Distribusi Indikator Kemiskinan dan IPM per Pulau
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(data=df, x='pulau', y='persentase_penduduk_miskin', palette='Set2', ax=axes[0])
axes[0].set_title('Distribusi Persentase Penduduk Miskin (%) per Pulau', fontsize=12, fontweight='bold')
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=30)

sns.boxplot(data=df, x='pulau', y='indeks_pembangunan_manusia_ipm', palette='Pastel1', ax=axes[1])
axes[1].set_title('Distribusi Indeks Pembangunan Manusia (IPM) per Pulau', fontsize=12, fontweight='bold')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=30)

plt.tight_layout()
plt.show()
```

```python
# 4. Heatmap Korelasi Matriks Indikator Sosioekonomi
plt.figure(figsize=(10, 8))
corr = df[cols_key].corr()
sns.heatmap(corr, annot=True, cmap='RdYlBu', vmin=-1, vmax=1, fmt='.2f', linewidths=0.5)
plt.title('Matriks Korelasi Indikator Kemiskinan, Infrastruktur & Bansos BPS', fontsize=13, fontweight='bold')
plt.show()
```
