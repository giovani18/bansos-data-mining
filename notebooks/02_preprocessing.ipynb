# 02. Preprocessing & Feature Engineering
## Penyusunan Indeks Kemiskinan Multidimensi & Normalisasi Fitur

Notebook ini memuat tahapan **Data Preparation**:
1. Imputasi dan deteksi outlier berbasis IQR.
2. Rekayasa fitur komposit: **Indeks Kemiskinan Multidimensi (IKM)** dan **Indeks Infrastruktur Dasar**.
3. Standardisasi fitur dengan `StandardScaler` untuk persiapan input model klasterisasi.

```python
import sys
sys.path.append('..')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.preprocessing import BPSDataPreprocessor
from src.features import compute_bps_regional_features

# 1. Load Data
df_raw = pd.read_csv('../data/raw/real_bps_kabupaten_kota.csv')

# 2. Preprocessing
preprocessor = BPSDataPreprocessor(scaler_type='standard')
df_clean = preprocessor.clean_and_impute(df_raw)
df_clean = preprocessor.handle_outliers_iqr(df_clean)

# 3. Feature Engineering
df_feat = compute_bps_regional_features(df_clean)
print(f'Total Kolom setelah Feature Engineering: {df_feat.shape[1]}')
df_feat[['kabupaten_kota', 'indeks_infrastruktur_dasar', 'indeks_kemiskinan_multidimensi_ikm', 'status_alokasi_bansos']].head()
```

```python
# 4. Visualisasi Distribusi Indeks Kemiskinan Multidimensi (IKM)
plt.figure(figsize=(10, 5))
sns.histplot(df_feat['indeks_kemiskinan_multidimensi_ikm'], kde=True, color='darkorange', bins=30)
plt.title('Distribusi Indeks Kemiskinan Multidimensi (IKM) 514 Kabupaten/Kota', fontsize=12, fontweight='bold')
plt.xlabel('Skor IKM (0: Paling Sejahtera, 100: Paling Miskin/Rentan)')
plt.show()
```

```python
# 5. Standardisasi Fitur untuk Klasterisasi
features_for_clustering = [
    'persentase_penduduk_miskin',
    'garis_kemiskinan',
    'indeks_kedalaman_p1',
    'indeks_keparahan_p2',
    'pengeluaran_per_kapita',
    'indeks_pembangunan_manusia_ipm',
    'persentase_sanitasi_layak',
    'persentase_air_minum_layak',
    'persentase_rumah_layak_huni',
    'indeks_infrastruktur_dasar',
    'indeks_kemiskinan_multidimensi_ikm'
]
X_scaled, df_scaled = preprocessor.scale_features(df_feat, features_for_clustering)
print('Matriks fitur siap klasterisasi dengan bentuk:', X_scaled.shape)
```
